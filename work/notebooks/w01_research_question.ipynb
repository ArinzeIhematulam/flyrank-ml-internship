## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Loading the starter dataset (30,000 pages), the overall declining rate is 54.2%. Pages matching the "stale (180+ days since update) and visible (500+ impressions)" pattern from ML-01 show a 94.1% declining rate — a 1.74x lift over baseline, suggesting the pattern is a real signal worth pursuing at scale. However, only 17 pages (0.1% of the starter dataset) match this exact threshold — too small a sample to treat as proven on its own. This is exactly why Lane 2 is worth the next 7 weeks: the full warehouse has ~520,000 content items and ~79M daily rows, which should surface enough matching pages to validate (or reject) this pattern with real statistical confidence, rather than 17 data points.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print(f"Total pages: {df.shape[0]}")

baseline_decline_rate = df["trend_direction"].str.lower().eq("down").mean()
print(f"Overall declining rate: {baseline_decline_rate:.1%}")

stale_visible = df[(df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)]
print(f"Stale + visible pages: {len(stale_visible)} ({len(stale_visible)/len(df):.1%} of total)")

decline_rate_stale_visible = stale_visible["trend_direction"].str.lower().eq("down").mean()
print(f"Declining rate among stale+visible pages: {decline_rate_stale_visible:.1%}")
print(f"Lift vs baseline: {decline_rate_stale_visible / baseline_decline_rate:.2f}x")
```
